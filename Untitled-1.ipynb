```python
def factorial(n):
    """Return n! using recursion."""
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n must be an integer.")
    if n < 0:
        raise ValueError("n must be non-negative.")
    if n == 0:  # 0! = 1
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    """Return the number of ways to choose r items from n, where order does not matter."""
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n must be an integer.")
    if not isinstance(r, int) or isinstance(r, bool):
        raise TypeError("r must be an integer.")
    if n < 0 or r < 0:
        raise ValueError("n and r must be non-negative.")
    if r > n:
        raise ValueError("r cannot be greater than n.")

    return factorial(n) // (factorial(r) * factorial(n - r))


# Test the valid case
n = 7
r = 2
print(f"Possible catalyst sets for n={n}, r={r}: {combination(n, r)}")

# Show the error message for the invalid case
n = 4
r = 6
try:
    print(f"Possible catalyst sets for n={n}, r={r}: {combination(n, r)}")
except ValueError as error:
    print(f"Invalid input for n={n}, r={r}: {error}")
```
